```python
import pandas as pd

df = pd.read_csv("gld_price_data.csv")
df.head()
```

## ⭐ PASO 3 — Limpieza y preparación del dataset
- Revisa tipos de datos
- Convierte fechas
- Ordena por fecha
- Maneja valores nulos

```python
# Revisar tipos de datos
df.info()

# Convertir columna de fechas
df["Date"] = pd.to_datetime(df["Date"])

# Ordenar por fecha
df = df.sort_values("Date")

# Revisar valores nulos
print(df.isna().sum())

# Manejar nulos: eliminar o rellenar
df = df.dropna()   # o df = df.fillna(method="ffill")

# Verificar
df.info()
df.head()
# TODO: escribir aquí tu proceso de limpieza
```

## ⭐ PASO 4 — Análisis financiero básico
Calcula:
- Rendimiento diario
- Rendimiento promedio
- Volatilidad
- Máx y mín históricos

```python
# TODO: calcular métricas financieras
# --- Cálculo de rendimiento diario (usando la columna GLD) ---
df["Daily_Return"] = df["GLD"].pct_change()

# --- Rendimiento promedio diario ---
rendimiento_promedio = df["Daily_Return"].mean()

# --- Volatilidad diaria ---
volatilidad = df["Daily_Return"].std()

# --- Máximo y mínimo histórico del precio GLD ---
precio_max = df["GLD"].max()
precio_min = df["GLD"].min()

# --- Mostrar resultados ---
print("📈 Rendimiento promedio diario:", rendimiento_promedio)
print("⚡ Volatilidad diaria:", volatilidad)
print("🔝 Máximo histórico GLD:", precio_max)
print("🔻 Mínimo histórico GLD:", precio_min)
```
