## ¿Qué es el Valor Presente Neto (VPN)?

El **Valor Presente Neto (VPN)** permite determinar si un proyecto de inversión es rentable.

- Si **VPN > 0** → el proyecto genera valor.
- Si **VPN < 0** → el proyecto no es rentable.

Fórmula:

$$VPN = \\sum \
rac{Flujo_t}{(1+r)^t} - Inversión\ Inicial$$

## Proyectos de inversión

Usaremos **diccionarios en Python** para representar cada proyecto.

```python
proyecto_A = {
    'nombre': 'Proyecto Solar',
    'inversion': 8000,
    'flujos': [2500, 3000, 3500, 4000],
    'tasa': 0.10
}

proyecto_B = {
    'nombre': 'Proyecto Tecnológico',
    'inversion': 10000,
    'flujos': [3000, 3500, 4200, 4500],
    'tasa': 0.10
}

proyecto_C = {
    'nombre': 'Proyecto Logístico',
    'inversion': 7000,
    'flujos': [1800, 2600, 3200, 3800],
    'tasa': 0.10
}

```

## Calcular el VPN usando `for`

```python
vpn = -proyecto_A['inversion']

for t, flujo in enumerate(proyecto_A['flujos']):
    vpn += flujo / (1 + proyecto_A['tasa'])**(t+1)

print('VPN del proyecto A:', round(vpn, 3))
```

## Decisión con `if`

```python
if vpn > 0:
    print('El proyecto es rentable')
else:
    print('El proyecto no es rentable')
```

## Analizar todos los proyectos

```python
proyectos = [proyecto_A, proyecto_B, proyecto_C]

for proyecto in proyectos:
    vpn = -proyecto['inversion']

    for t, flujo in enumerate(proyecto['flujos']):
        vpn += flujo / (1 + proyecto['tasa'])**(t+1)

    print(proyecto['nombre'], 'VPN:', round(vpn, 2))
```

## Simulación de tasas con `while`

```python
tasa = 0.05

while tasa <= 0.20:
    vpn = -proyecto_A['inversion']

    for t, flujo in enumerate(proyecto_A['flujos']):
        vpn += flujo / (1 + tasa)**(t+1)

    print('Tasa:', tasa, 'VPN:', round(vpn, 2))

    tasa += 0.05
```

```python
# Solucion punto 1 se calcula las VPN de los tres proyectos                   # Explicacion 1: En este punto se lista la informacion de los 3 proyectos
proyecto_A = {
    'nombre': 'Proyecto Solar',
    'inversion': 8000,
    'flujos': [2500, 3000, 3500, 4000],
    'tasa': 0.10
}

proyecto_B = {
    'nombre': 'Proyecto Tecnológico',
    'inversion': 10000,
    'flujos': [3000, 3500, 4200, 4500],
    'tasa': 0.10
}

proyecto_C = {
    'nombre': 'Proyecto Logístico',
    'inversion': 7000,
    'flujos': [1800, 2600, 3200, 3800],
    'tasa': 0.10
}

proyectos = [proyecto_A, proyecto_B, proyecto_C]                             # Explicacion 2: Se crea una lista con los proyectos A,B,C llamada proyectos


for proyecto in proyectos:                                                   # Explicacion 3: Se crea un bucle for llamada proyectos, en cada iteración la variable proyecto toma uno de los elementos de la lista.
    vpn = -proyecto['inversion']                                             # Explicacion 4: Se esta inicializando la variable vpn con el valor negatico de la inversion inical del proyecto (porque?En el cálculo del Valor Presente Neto (VPN), la inversión inicial se considera un egreso (un desembolso de dinero), por eso se coloca con signo negativo.)

    for t, flujo in enumerate(proyecto['flujos']):                           # Explicacion 5: En este paso se realiza un bucle compuesto en el cual tenemos un for dentro de otro for el cual recorre la lista de flujos de un proyecto - enumerate() es una función que devuelve dos cosas en cada iteración: t: el índice (posición en la lista, empezando en 0) y flujo: el valor del flujo en esa posición


        vpn += flujo / (1 + proyecto['tasa'])**(t+1)                         # Explicacion 6: Se suma el acumulado de los resultados de la operacion realizada anteriormnente para el calculo de la vpn

    print(proyecto['nombre'], 'VPN:', round(vpn, 2))                         # Explicacion 7: Se realiza la impresion de la informacion de los proyectos ABC con su respectivo valor de la vpn
```

```python
# Solucion punto 2, determinar cual proyecto tiene la VPN mas alto

proyectos = [proyecto_A, proyecto_B, proyecto_C]              # Explicacion 1: En este punto se lista la informacion de los 3 proyectos
mejor_proyecto = None                                         # Explicacion 2: Se inicializa la variable sin valor, esto indica que al inicio todavia no se ha encontrado ningun proyecto como "el mejor"
mejor_vpn = float('-inf')                                     # Explicacion 3: se inicializa la variable con el valor menos infinito, ya que cualquier vpn calculado sera mayor que -inf

for proyecto in proyectos:                                    # Explicacion 4: En estos dos bucles for se realiza la operacion del calculo de las vpn que se explico con anterioridad en el punto 1 y se imprimen sus valores
    vpn = -proyecto['inversion']
    for t, flujo in enumerate(proyecto['flujos']):
        vpn += flujo / (1 + proyecto['tasa'])**(t+1)

    print(proyecto['nombre'], 'VPN:', round(vpn, 3))

    # Verificamos si este proyecto supera al mejor encontrado
    if vpn > mejor_vpn:                                      # Explicacion 5: En este paso se hace la comparacion si el VPN calculado para el proyecto actual vpn es mayor que el mejor VPN encontrado hasta el momento mejor_vpn
        mejor_vpn = vpn                                      # Explicacion 6: Si la condicion anterior es verdadera, se actualiza la variable mejor_vpn con el nuevo valor, esto significa que la variable mejor_vpn es el que tiene el mejor proyecto al momento
        mejor_proyecto = proyecto['nombre']                  # Explicacion 7: Ademas se guarda el nombre del proyecto que se guardo en la variable anterior mejor_vpn

print("\nEl proyecto con el VPN más alto es:", mejor_proyecto, "con un VPN de", round(mejor_vpn, 3)) # Explicacion 7: En este paso se realiza la impresion de la informacion del mejor proyecto con su vpn
```
